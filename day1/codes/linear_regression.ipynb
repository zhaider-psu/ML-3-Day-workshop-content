```python

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.datasets import load_diabetes
import matplotlib.pyplot as plt
import numpy as np

def regression_demo_linear():
    """
    Regression demo on diabetes dataset using Linear Regression.
    """
    # 1. Load Data
    # For visualization purposes, we'll use only one feature (BMI)
    X, y = load_diabetes(return_X_y=True)
    print(f"Shape of X: {X.shape}")
    print(f"Shape of y: {y.shape}")
    X = X[:, np.newaxis, 2]  # Use only BMI feature

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

    # 2. Train Model
    model = LinearRegression()
    model.fit(X_train, y_train)          # Train

    # 3. Evaluate
    y_pred = model.predict(X_test)       # Predict

    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    print(f"LinearRegression: MSE={mse:.2f}, R2={r2:.3f}")

    # 4. Visualize
    plt.scatter(X_test, y_test, color='black', label='Actual Data')
    plt.plot(X_test, y_pred, color='blue', linewidth=3, label='Learned Line')
    plt.xlabel('BMI (Normalized)')
    plt.ylabel('Disease Progression')
    plt.title('Linear Regression Model')
    plt.legend()
    plt.show()

# calling the function
regression_demo_linear()
```
