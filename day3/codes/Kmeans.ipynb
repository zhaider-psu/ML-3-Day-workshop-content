```python
from sklearn.cluster import KMeans, DBSCAN, SpectralClustering
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
import matplotlib.pyplot as plt

def clustering_demo():
    # Generate synthetic data (two moons)
    X, y = datasets.make_moons(n_samples=400, noise=0.08, random_state=0)
    scaler = StandardScaler()
    X_s = scaler.fit_transform(X)

    # KMeans
    kmeans = KMeans(n_clusters=2, random_state=0)
    labels_km = kmeans.fit_predict(X_s)

    # Gaussian Mixture Models (GMM)
    gmm = GaussianMixture(n_components=2, random_state=0)
    labels_gmm = gmm.fit_predict(X_s)

    # Plotting (helper function)
    plt.scatter(X_s[:, 0], X_s[:, 1], c=labels_km, cmap='viridis')
    plt.title("KMeans on two moons")
    plt.show()
clustering_demo()
```
