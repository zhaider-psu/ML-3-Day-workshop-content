```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

def pca_demo_plot():
    """
    PCA demo on Iris dataset.
    """
    # 1. Load Data
    data = load_iris()
    X = data.data
    y = data.target

    # 2. Scale Features (CRITICAL for PCA!)
    # PCA is sensitive to scale, so we must standardize first.
    scaler = StandardScaler()
    X_s = scaler.fit_transform(X)

    # 3. Apply PCA
    pca = PCA(n_components=2)
    Z = pca.fit_transform(X_s)

    # 4. Check results
    print("Explained variance ratio:", pca.explained_variance_ratio_)
    # e.g., [0.72, 0.23] means 1st PC explains 72% of variance, 2nd explains 23%.

    # 5. Plot
    plt.scatter(Z[:, 0], Z[:, 1], c=y, cmap='viridis')
    plt.xlabel("First Principal Component")
    plt.ylabel("Second Principal Component")
    plt.title("Iris Dataset Projected to 2D")
    plt.show()
pca_demo_plot()
```
